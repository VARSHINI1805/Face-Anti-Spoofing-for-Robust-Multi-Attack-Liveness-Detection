```python
import os
import json
from collections import Counter

# Load labels
with open("CelebA_Spoof_/CelebA_Spoof/metas/intra_test/train_label.json", "r") as f:
    data = json.load(f)

# Root folder containing Data/
root = "CelebA_Spoof_/CelebA_Spoof"

attack_names = {
    0: "Live",
    1: "Photo",
    2: "Poster",
    3: "A4",
    4: "Face Mask",
    5: "Upper Body Mask",
    6: "Region Mask",
    7: "PC",
    8: "Pad",
    9: "Phone",
    10: "3D Mask"
}

counter = Counter()

for img_path, labels in data.items():

    # Complete image path
    full_path = os.path.join(root, img_path)

    # Count only if image exists
    if os.path.isfile(full_path):
        attack = labels[40]
        counter[attack] += 1

print("Image Count for Each Attack\n")

total = 0
for cls in sorted(counter.keys()):
    print(f"{attack_names[cls]:20s} : {counter[cls]}")
    total += counter[cls]

print("\nTotal Images :", total)
```

```python
import os
from PIL import Image
import json

with open("CelebA_Spoof_/CelebA_Spoof/metas/intra_test/train_label.json") as f:
    data = json.load(f)

root = "CelebA_Spoof_/CelebA_Spoof"

widths = []
heights = []

count = 0

for img_path in data.keys():

    full_path = os.path.join(root, img_path)

    if os.path.isfile(full_path):

        img = Image.open(full_path)

        w, h = img.size

        widths.append(w)
        heights.append(h)

        count += 1

        if count == 5000:
            break

print("Average Width :", sum(widths)/len(widths))
print("Average Height:", sum(heights)/len(heights))
print("Minimum Width :", min(widths))
print("Maximum Width :", max(widths))
print("Minimum Height:", min(heights))
print("Maximum Height:", max(heights))
```

```python
import os
from collections import Counter

root = "CelebA_Spoof_/CelebA_Spoof"

format_counter = Counter()

for img_path in data.keys():

    full_path = os.path.join(root, img_path)

    if os.path.isfile(full_path):
        ext = os.path.splitext(full_path)[1].lower()
        format_counter[ext] += 1

print("=" * 40)
print("IMAGE FORMAT STATISTICS")
print("=" * 40)

total = sum(format_counter.values())

for ext, count in sorted(format_counter.items()):
    print(f"{ext:8s} : {count:,} images")

print("-" * 40)
print(f"Total Images : {total:,}")
```

```python
import os
import json
import random
import matplotlib.pyplot as plt
from PIL import Image

# Dataset root
root = "CelebA_Spoof_/CelebA_Spoof"

# Load labels
with open(os.path.join(root, "metas/intra_test/train_label.json"), "r") as f:
    data = json.load(f)

# Attack names
attack_names = {
    0: "Live",
    1: "Photo",
    2: "Poster",
    3: "A4",
    4: "Face Mask",
    5: "Upper Body Mask",
    6: "Region Mask",
    7: "PC",
    8: "Pad",
    9: "Phone",
    10: "3D Mask"
}

# Group images by attack class
attack_images = {i: [] for i in range(11)}

for img_path, labels in data.items():
    attack = labels[40]
    full_path = os.path.join(root, img_path)

    if os.path.exists(full_path):
        attack_images[attack].append(full_path)

plt.figure(figsize=(18,12))

for attack in range(11):

    # Randomly choose one image
    img_path = random.choice(attack_images[attack])

    img = Image.open(img_path)

    plt.subplot(3,4,attack+1)
    plt.imshow(img)
    plt.title(attack_names[attack], fontsize=12)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os
import json
import random
import matplotlib.pyplot as plt
from PIL import Image

root = "CelebA_Spoof_/CelebA_Spoof"

with open(os.path.join(root, "metas/intra_test/train_label.json")) as f:
    data = json.load(f)

attack_names = {
    0:"Live",
    1:"Photo",
    2:"Poster",
    3:"A4",
    4:"Face Mask",
    5:"Upper Body Mask",
    6:"Region Mask",
    7:"PC",
    8:"Pad",
    9:"Phone",
    10:"3D Mask"
}

attack_images = {i: [] for i in range(11)}

for img_path, labels in data.items():
    attack = labels[40]
    full_path = os.path.join(root, img_path)
    if os.path.exists(full_path):
        attack_images[attack].append(full_path)

plt.figure(figsize=(18,30))

plot = 1

for attack in range(11):

    samples = random.sample(
        attack_images[attack],
        min(5, len(attack_images[attack]))
    )

    for img_path in samples:

        img = Image.open(img_path)

        plt.subplot(11,5,plot)
        plt.imshow(img)
        plt.title(attack_names[attack], fontsize=8)
        plt.axis("off")

        plot += 1

plt.tight_layout()
plt.show()
```

```python
import os

root = "CelebA_Spoof_/CelebA_Spoof/Data"

def count_images(folder):

    count = 0

    for path, dirs, files in os.walk(folder):

        for file in files:

            if file.lower().endswith((".jpg", ".png", ".jpeg")):
                count += 1

    return count

train = count_images(os.path.join(root, "train"))
test = count_images(os.path.join(root, "test"))

print("="*45)
print("FINAL DATASET DISTRIBUTION")
print("="*45)
print(f"Training Images : {train:,}")
print(f"Testing Images  : {test:,}")
print(f"Total Images    : {train+test:,}")
```

```python
import os
import json
from PIL import Image

root = "CelebA_Spoof_/CelebA_Spoof"

with open(os.path.join(root, "metas/intra_test/train_label.json")) as f:
    data = json.load(f)

missing = 0
corrupted = 0

for img_path in list(data.keys())[:10000]:

    full_path = os.path.join(root, img_path)

    if not os.path.exists(full_path):
        missing += 1
        continue

    try:
        img = Image.open(full_path)
        img.verify()
    except:
        corrupted += 1

print("="*45)
print("DATASET QUALITY REPORT")
print("="*45)
print("Checked Images :", 10000)
print("Missing Images :", missing)
print("Corrupted Images :", corrupted)
```
